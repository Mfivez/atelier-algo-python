# 1 — Comprendre la complexité pas à pas


## Fil conducteur

Tout le chapitre répond à une seule question :

> Quand la quantité de données augmente, comment le travail du programme évolue-t-il ?

Nous utiliserons surtout de petites listes afin de voir chaque opération. Le
raisonnement avancera toujours dans le même ordre :

1. définir la taille de l'entrée et lui donner le nom `n` ;
2. suivre une boucle case par case et compter ses opérations ;
3. comparer un accès direct, deux boucles successives et deux boucles imbriquées ;
4. observer ce qui change lorsque `n` devient beaucoup plus grand ;
5. repérer le travail parfois caché dans une instruction Python ;
6. séparer le temps de calcul de la mémoire utilisée ;
7. terminer par la correction : expliquer pourquoi une boucle donne le bon résultat.

À chaque exemple, nous poserons donc les mêmes questions : que représente `n` ?
Quelle action est répétée ? Combien de fois ? Quelle information faut-il garder ?
Cette répétition rend le fil logique visible du premier comptage jusqu'à la
justification finale.




*Chaque étape répond à une question précise avant de passer à la suivante.*


## 1. La complexité ne veut pas dire « compliqué »

Ici, la **complexité** décrit l'évolution des ressources nécessaires à un
algorithme. Une ressource peut être :

- le **temps de calcul**, représenté par un nombre d'opérations ;
- la **mémoire**, représentée par la quantité d'informations conservées.

Nous ne commençons pas par mesurer des secondes. Deux ordinateurs peuvent donner
des durées différentes pour le même code. En revanche, ils doivent tous les deux
traiter les éléments nécessaires. Compter ce travail permet de prévoir ce qui
arrivera lorsque l'entrée deviendra beaucoup plus grande.

### Exemple illustré — observer une croissance

| Taille de la liste | Lire le premier élément | Additionner toute la liste |
|---:|---:|---:|
| 3 | 1 lecture | 3 lectures |
| 6 | 1 lecture | 6 lectures |

La liste double. Le premier travail reste identique ; le second double. La
complexité décrit cette évolution, pas la difficulté de lecture du code.


## 2. La taille de l'entrée : n

Le symbole **n**, lu « enne », représente la taille de l'entrée.

- Pour une liste, n est généralement son nombre d'éléments.
- Pour une chaîne, n peut être son nombre de caractères.
- Plus tard, un graphe demandera deux tailles : le nombre de sommets et le nombre
  d'arêtes.

La taille n'est pas la valeur des données. La liste [1, 2, 3] et la liste
[1000000, -8, 42] ont toutes les deux une taille n = 3.

Avant toute analyse, pose donc la question : **qu'est-ce qui peut grandir ?**

### Exemple illustré — donner un sens à `n`

| Entrée | Ce que compte `n` | Valeur de `n` |
|---|---|---:|
| `[8, 3, 1]` | éléments de la liste | 3 |
| `"chat"` | caractères de la chaîne | 4 |
| `A → B → C` | sommets du petit graphe | 3 |

Avant tout calcul de coût, il faut pouvoir terminer la phrase : « ici, `n`
représente… ».


## 3. Premier exemple : parcourir une liste

La fonction suivante compte les valeurs strictement positives. Lis d'abord le
code sans parler de Big O. Pour chaque élément, elle pose une fois la question
« cette valeur est-elle positive ? ».

### Exemple illustré — suivre la boucle case par case

Pour `[4, -1, 2]` :

| Indice | Valeur lue | Est-elle positive ? | Compteur après le test |
|---:|---:|:---:|---:|
| 0 | 4 | oui | 1 |
| 1 | -1 | non | 1 |
| 2 | 2 | oui | 2 |

Une ligne du tableau correspond à un tour de boucle.


````python
def compter_positifs(valeurs):
    compteur = 0
    for valeur in valeurs:
        if valeur > 0:
            compteur += 1
    return compteur

assert compter_positifs([-2, 0, 3, 4]) == 2
assert compter_positifs([]) == 0
````

La boucle fait un tour par élément :

| taille n | tours de boucle | comparaisons valeur > 0 |
|---:|---:|---:|
| 0 | 0 | 0 |
| 1 | 1 | 1 |
| 3 | 3 | 3 |
| 100 | 100 | 100 |

Si la liste devient deux fois plus longue, le nombre de comparaisons devient
deux fois plus grand. On dit que le travail grandit **linéairement** :
il suit la même progression que n.


### Lecture visuelle d’un parcours linéaire

Pour la liste `[4, -2, 7, 0]`, chaque case provoque exactement une comparaison :

| Indice | 0 | 1 | 2 | 3 |
|---:|:---:|:---:|:---:|:---:|
| Valeur | 4 | -2 | 7 | 0 |
| Test `> 0` | vrai | faux | vrai | faux |
| Travail effectué | 1 test | 1 test | 1 test | 1 test |

Quatre cases donnent quatre tests. Avec `n` cases, la même relation donne `n`
tests : c’est cette correspondance, et non la durée en secondes, que résume O(n).

## 4. O(n) : un travail qui suit la taille

La notation **O(n)**, lue « grand O de n », signifie ici :

> Quand n grandit, le nombre d'opérations grandit au plus comme une quantité
> proportionnelle à n.

O(n) ne donne pas le nombre exact d'opérations. Une boucle peut effectuer trois
petites actions par élément et une autre dix. Elles restent toutes les deux O(n),
car doubler n multiplie leur travail par environ deux.

On ignore donc les nombres fixes pour observer la tendance. n, 2n et 5n + 20
ont tous une croissance linéaire. Cela ne signifie pas que leurs durées réelles
sont identiques ; cela signifie qu'elles changent de la même manière lorsque n
devient grand.

### Exemple illustré — reconnaître une croissance linéaire

```text
n = 2  → ■ ■
n = 4  → ■ ■ ■ ■
n = 8  → ■ ■ ■ ■ ■ ■ ■ ■
```

Chaque carré représente une opération. Multiplier `n` par 2 multiplie aussi le
travail par 2 : c’est le comportement résumé par O(n).


## 5. O(1) : un travail indépendant de n

Considère maintenant l'accès au premier élément d'une liste non vide.

### Exemple illustré — une seule case suffit

```text
liste courte : [●][ ][ ]          → lire la case 0
liste longue : [●][ ][ ][ ]…[ ]   → lire la case 0
```

Le point noir est l’unique case consultée. Le nombre total de cases ne change pas
le travail : on parle d’un coût O(1).


````python
def premier(valeurs):
    if not valeurs:
        raise ValueError("liste vide")
    return valeurs[0]

assert premier([8]) == 8
assert premier([8, 4, 2, 1]) == 8
````

Que la liste contienne 1 ou 1 000 000 d'éléments, on lit une seule position.
Le travail ne dépend pas de n : on le note **O(1)**, ou coût constant.

O(1) ne veut pas dire « instantané » ni « exactement une instruction ».
Cela veut seulement dire que la quantité de travail ne grandit pas avec
la taille étudiée.

Nous savons maintenant reconnaître deux situations : consulter une position
connue est constant ; parcourir tous les éléments est linéaire.


## 6. Deux boucles l'une après l'autre

Supposons qu'une première boucle parcoure n éléments, puis qu'une seconde boucle
parcoure encore n éléments. Le travail total est n + n = 2n.

Comme le nombre fixe 2 ne change pas la manière dont le travail grandit, le total
reste O(n). **Deux boucles ne donnent donc pas automatiquement O(n²).**

La bonne question est : les boucles se suivent-elles, ou l'une recommence-t-elle
entièrement pour chaque tour de l'autre ?

### Exemple illustré — deux parcours successifs s’additionnent

```text
minimum :  → [3] → [1] → [8]     3 lectures
maximum :  → [3] → [1] → [8]     3 lectures
total :                              6 lectures
```

Pour `n` éléments, cela donne `n + n = 2n`. Les parcours ne se multiplient pas,
car le second commence après la fin du premier.




*On additionne les parcours successifs ; on multiplie lorsque le parcours intérieur recommence entièrement.*


````python
def minimum_et_maximum(valeurs):
    if not valeurs:
        raise ValueError("liste vide")
    minimum = valeurs[0]
    for valeur in valeurs:
        if valeur < minimum:
            minimum = valeur
    maximum = valeurs[0]
    for valeur in valeurs:
        if valeur > maximum:
            maximum = valeur
    return minimum, maximum

assert minimum_et_maximum([4, -2, 9, 1]) == (-2, 9)
````

## 7. Une boucle dans une boucle : O(n²)

Pour comparer chaque élément à chaque autre, la boucle intérieure recommence n
fois pour chacun des n tours extérieurs. Le nombre de couples est n × n = n².
On lit n² « n au carré ».

### Exemple illustré — une grille de couples

Avec trois valeurs, deux boucles imbriquées visitent une grille de `3 × 3` cases :

| extérieur / intérieur | 0 | 1 | 2 |
|---:|:---:|:---:|:---:|
| 0 | ■ | ■ | ■ |
| 1 | ■ | ■ | ■ |
| 2 | ■ | ■ | ■ |

Les neuf carrés représentent `3²` couples. Avec `n` valeurs, la grille contient
`n²` cases.


````python
def nombre_de_couples(valeurs):
    compteur = 0
    for _ in valeurs:
        for _ in valeurs:
            compteur += 1
    return compteur

assert nombre_de_couples([]) == 0
assert nombre_de_couples([1, 2, 3]) == 9
````

| taille n | couples n² |
|---:|---:|
| 1 | 1 |
| 2 | 4 |
| 10 | 100 |
| 100 | 10 000 |

Si n double, n² est multiplié par quatre. Cette croissance **quadratique**,
notée O(n²), devient donc beaucoup plus vite coûteuse qu'une croissance linéaire.

Nous avons construit les trois formes essentielles du début du cours :
O(1), O(n) et O(n²). La suite consiste à repérer le travail moins visible.




*Le dessin compare des nombres d’opérations : O(1) reste stable, O(n) suit n et O(n²) accélère.*

## 8. Le travail caché dans une instruction Python

Une ligne de code peut déclencher plusieurs opérations :

- <code>len(liste)</code> lit une taille déjà conservée : O(1) ;
- <code>x in liste</code> peut chercher jusqu'au dernier élément : O(n) ;
- <code>liste[a:b]</code> crée une **tranche**, c'est-à-dire une nouvelle liste
  contenant les éléments choisis : O(k) pour k éléments copiés ;
- <code>sum(liste)</code> parcourt les éléments : O(n) ;
- <code>sorted(liste)</code> crée une copie triée : O(n log n).

Le terme **log n** sera expliqué plus loin comme une progression obtenue en
réduisant fortement la zone à chaque étape. Pour l'instant, retiens seulement
qu'un tri général est plus coûteux qu'un parcours simple, mais beaucoup moins
qu'une comparaison de toutes les paires.

Analyser un programme demande donc de regarder ce que font les opérations,
pas seulement de compter ses lignes.

### Exemple illustré — une ligne de Python peut cacher une boucle

```text
7 in [2, 4, 7, 9]
      2 ?  →  4 ?  →  7 ✓
```

L’écriture tient sur une ligne, mais Python a comparé plusieurs éléments. De même,
`valeurs[:]` paraît court tout en copiant `n` éléments.


## 9. Le pire cas

Une recherche peut s'arrêter tôt si la valeur se trouve au début. Si elle est
absente, il faut parcourir toute la liste.

### Exemple illustré — comparer les cas d’une recherche

Recherche d'une cible dans une liste de quatre éléments :

| Situation | Comparaisons |
|---|---:|
| `7` dans `[7, 2, 4, 9]` | 1 |
| `7` dans `[2, 4, 9, 7]` | 4 |
| `7` dans `[2, 4, 9, 8]` | 4 |

Le pire cas retient la ligne la plus coûteuse : ici, jusqu’à `n` comparaisons.


````python
def contient(valeurs, cible):
    for valeur in valeurs:
        if valeur == cible:
            return True
    return False

assert contient([5, 8, 2], 5)
assert not contient([5, 8, 2], 7)
````

Le **meilleur cas** est la situation la plus favorable : ici, une comparaison.
Le **pire cas** est la situation qui demande le plus de travail : ici, n
comparaisons. Sauf indication contraire, ce cours annonce généralement le pire
cas, car il donne une garantie.

Un **cas moyen** demanderait de préciser quelles entrées sont probables. Sans
cette hypothèse, le mot « moyen » n'a pas de signification assez précise.


## 10. Temps et mémoire sont deux questions différentes

Jusqu'ici, nous avons compté le travail. Un algorithme peut aussi conserver des
informations. On distingue :

- **l'entrée**, qui existait avant l'appel ;
- **la mémoire de travail**, créée pour calculer ;
- **la sortie**, créée pour rendre le résultat.

Une variable compteur prend une quantité fixe de mémoire : O(1). Une nouvelle
liste qui contient n résultats prend O(n). Dire simplement « mémoire O(1) »
serait trompeur si la fonction construit une grande sortie.

### Exemple illustré — même temps, mémoire différente

| Fonction | Parcours | Mémoire créée pendant le calcul |
|---|---:|---:|
| somme des valeurs | `n` lectures | un total |
| liste des carrés | `n` lectures | `n` résultats |

Les deux temps sont linéaires. La première fonction ajoute une mémoire fixe ; la
seconde construit une sortie qui grandit avec `n`.




*Parcourir les mêmes données peut demander le même temps tout en créant une quantité de mémoire différente.*

````python
def carres(valeurs):
    resultat = []
    for valeur in valeurs:
        resultat.append(valeur * valeur)
    return resultat

assert carres([2, 3, 4]) == [4, 9, 16]
````

La fonction effectue un parcours : O(n) en temps. Sa liste résultat contient n
éléments : O(n) pour la sortie. En dehors de cette sortie, elle n'utilise qu'une
référence de boucle et quelques variables : O(1) mémoire de travail.

Cette séparation deviendra importante pour le backtracking : produire des
milliers de solutions oblige à stocker ou transmettre ces milliers de solutions.


## 11. Une carte pour les chapitres suivants

Les trois formes à maîtriser maintenant restent O(1), O(n) et O(n²). Les autres
notations ci-dessous servent seulement à reconnaître ce qui arrivera plus tard.

- **O(log n)**, lu « O de logarithme de n » : on réduit fortement la zone à
  chaque étape. Si on divise par deux une taille 16, on obtient
  16 → 8 → 4 → 2 → 1 : seulement quatre réductions. Doubler la taille ajoute
  approximativement une étape.
- **O(n log n)** : on combine n éléments avec un nombre logarithmique de niveaux
  de travail. C'est le coût courant d'un tri général efficace.
- **O(2ⁿ)**, lu « O de deux puissance n » : chaque nouvel élément double le
  nombre de possibilités. Pour 1, 2 et 3 choix binaires, on obtient 2, 4 et
  8 possibilités.
- **O(n!)**, lu « O de factorielle n » : le symbole n! signifie
  n × (n−1) × ... × 2 × 1. Il compte notamment tous les ordres possibles de
  n éléments.

Les chapitres concernés reconstruiront ces coûts à partir de leurs arbres ou de
leurs opérations. Cette carte n'est pas une liste à apprendre par cœur.

### Exemple illustré — une information mène à une technique

```text
Question : « ai-je déjà vu 7 ? »
        ↓ information à conserver
ensemble des valeurs vues
        ↓ structure adaptée
set Python
```

Les chapitres suivants suivront toujours ce chemin : partir de la question, choisir
l’information utile, puis seulement choisir une structure ou une technique.


## Transition : rapide ne suffit pas

Nous savons maintenant décrire le coût d'un programme. Mais un programme très
rapide qui rend une mauvaise réponse ne sert à rien. La deuxième moitié du
chapitre répond donc à une nouvelle question, directement liée à la première :

> Comment expliquer que la boucle calcule bien ce qu'elle promet ?

Nous allons reprendre une boucle simple et suivre le sens de sa variable.


## 12. Un invariant expliqué simplement

Un **invariant** est une phrase qui reste vraie à chaque étape de la boucle,
même si les valeurs des variables changent.

Pour une somme, la phrase est :

> total est la somme des éléments déjà parcourus.

Sur [2, -1, 4] :

| éléments déjà parcourus | total | phrase vérifiée |
|---|---:|---|
| aucun | 0 | la somme de rien vaut 0 |
| 2 | 2 | total vaut 2 |
| 2, -1 | 1 | total vaut 2 + (-1) |
| 2, -1, 4 | 5 | total vaut 2 + (-1) + 4 |

Le mot invariant ne signifie pas que <code>total</code> ne change jamais.
C'est le **sens** de <code>total</code> qui reste le même.

### Exemple illustré — une phrase vraie après chaque tour

Pour additionner `[2, 5, 1]` :

| Après avoir lu… | `total` | Phrase qui reste vraie |
|---|---:|---|
| rien | 0 | somme de la partie déjà lue |
| 2 | 2 | somme de `[2]` |
| 2, 5 | 7 | somme de `[2, 5]` |
| 2, 5, 1 | 8 | somme de toute la liste |

Cette phrase stable est l’invariant de la boucle.


````python
def somme(valeurs):
    total = 0
    for valeur in valeurs:
        total += valeur
    return total

assert somme([2, -1, 4]) == 5
assert somme([]) == 0
````

## 13. Justifier la boucle en quatre petites étapes

1. **Au départ** : aucun élément n'est parcouru et <code>total</code> vaut 0.
   L'invariant est vrai.
2. **Pendant un tour** : on ajoute exactement le nouvel élément. Le total devient
   donc la somme de tous les éléments parcourus jusque-là. L'invariant reste vrai.
3. **Pourquoi la boucle finit** : <code>for</code> avance dans une liste finie,
   un élément après l'autre.
4. **À la fin** : tous les éléments sont parcourus. L'invariant dit alors que
   <code>total</code> est la somme de toute la liste.

Les noms plus formels de ces étapes sont **initialisation**, **conservation**,
**terminaison** et **conclusion**. Les mots viennent après le raisonnement :
ils servent seulement à le résumer.

### Exemple illustré — les quatre moments d’une justification

| Moment | Question | Réponse pour une somme |
|---|---|---|
| initialisation | vrai avant la boucle ? | 0 est la somme de la partie vide |
| conservation | reste vrai après un tour ? | on ajoute exactement la valeur lue |
| terminaison | pourquoi la boucle finit-elle ? | l’indice avance vers la fin |
| conclusion | que sait-on à la fin ? | toute la liste a été additionnée |

Chaque ligne répond à une question différente ; aucune ne remplace les autres.


## 14. Tests et justification se complètent

Un test exécute le programme sur une entrée précise et compare le résultat à ce
qui est attendu. Il peut révéler une erreur. Mais même 10 000 tests ne couvrent
pas toutes les listes possibles.

La justification explique pourquoi toutes les entrées respectant le contrat
sont traitées correctement. Elle peut elle-même contenir une erreur de
raisonnement. En pratique :

- les tests confrontent le code à des exemples variés ;
- l'invariant et la terminaison expliquent le cas général.

On ne choisit pas entre les deux : on utilise les deux.

### Exemple illustré — tests et justification se complètent

| Outil | Ce qu’il apporte | Ce qu’il ne garantit pas seul |
|---|---|---|
| test sur `[]`, `[5]`, `[2, -1]` | révèle des erreurs concrètes | tous les futurs cas |
| justification de la boucle | raisonne pour toute taille `n` | absence d’erreur de frappe |

Les tests inspectent des points précis ; la justification relie tous les points par
un raisonnement général.
