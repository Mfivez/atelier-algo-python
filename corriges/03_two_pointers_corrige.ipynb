# 3 — Two pointers : déplacer deux indices avec une raison


## Fil conducteur

Le chapitre 2 gagnait du temps en mémorisant des informations. Ici, nous utilisons
une autre source d'information : l'ordre des valeurs. Sur une liste triée, une
comparaison permet d'éliminer sûrement une extrémité. Nous apprendrons d'abord

## 1. Deux pointeurs sont deux positions
Un pointeur est ici un **indice Python**, pas une adresse mémoire manipulée à la main.
On peut partir des deux extrémités, ou utiliser un indice de lecture et un indice
d'écriture. La fenêtre glissante appartient à cette famille de techniques,
mais toutes les techniques à deux indices ne décrivent pas une fenêtre.

### Exemple illustré — deux positions, deux rôles

```text
valeurs : [1][3][4][8][10]
           ↑           ↑
        gauche       droite
```

Les pointeurs sont ici de simples indices. Ils indiquent quelles cases sont encore
utiles au raisonnement.

Le schéma suivant distingue la variable, le numéro d’indice qu’elle contient et la case qu’elle désigne.




*`gauche` et `droite` stockent des indices : modifier leur valeur change la case observée, pas la liste.*


### Trois familles de déplacements

« Deux pointeurs » ne désigne pas un algorithme unique.

- **Extrémités opposées** : les deux indices se rapprochent grâce à un ordre ou à
  une symétrie, comme pour une somme dans un tableau trié ou un palindrome.
- **Lecture/écriture** : l'indice de lecture inspecte chaque valeur ; l'indice
  d'écriture désigne la prochaine position valide. L'invariant porte alors sur le
  préfixe déjà compacté.
- **Vitesses différentes** : un indice rapide et un indice lent peuvent repérer un
  cycle ou un milieu. Cette variante demande une preuve spécifique et n'est pas
  utilisée dans les exercices de ce pack.

Dans tous les cas, chaque déplacement doit avoir un sens logique. Compter deux
variables nommées `gauche` et `droite` ne garantit ni la correction ni un coût
linéaire.




*Lecture inspecte toutes les cases ; écriture délimite le préfixe déjà compacté.*

## 2. Trouver une paire de somme donnée
Dans une liste triée `[1,2,4,7,11]`, chercher une somme 9.
Au départ : 1+11=12, trop grand ; réduire la valeur de droite. 1+7=8,
trop petit ; augmenter la valeur de gauche. 2+7=9, trouvé.
Pourquoi peut-on éliminer une position ? Si le plus petit restant additionné
au plus grand est trop petit, l'additionner à n'importe quel autre restant
sera encore trop petit. Le tri donne cette certitude.

### Exemple illustré — la somme décide du mouvement

Pour une cible égale à 9 :

```text
1 + 10 = 11  → trop grand → droite recule
1 +  8 =  9  → trouvé
```

Comme la liste est triée, reculer `droite` est une élimination justifiée, pas un
choix au hasard.




*Chaque comparaison élimine une extrémité grâce à l’ordre de la liste.*


## 3. Invariant et terminaison
Si une paire existe et n'a pas été trouvée, une paire reste dans la zone entre
les indices. À chaque tour, une borne se rapproche de l'autre : au plus n−1
mouvements. On utilise `gauche < droite` pour ne pas employer deux fois le même
élément. Une valeur peut apparaître deux fois à des indices différents.

### Exemple illustré — l’intervalle utile rétrécit

| Étape | Zone encore possible | Taille |
|---:|---|---:|
| 0 | indices 0 à 5 | 6 |
| 1 | indices 1 à 5 | 5 |
| 2 | indices 1 à 4 | 4 |

L’invariant dit que toute paire encore possible se trouve dans la zone. La taille
diminue à chaque tour : la boucle finit.


### Preuve d'élimination

Supposons la zone restante triée entre `gauche` et `droite`.

- Si `nums[gauche] + nums[droite] < cible`, même la plus grande valeur restante
  ne suffit pas avec `nums[gauche]`. Aucune solution n'utilise donc cet indice :
  avancer `gauche` est sûr.
- Si la somme est trop grande, même la plus petite valeur restante produit une
  somme trop grande avec `nums[droite]`. Reculer `droite` est sûr.
- Si les valeurs sont égales à la cible, les deux indices sont distincts grâce à
  la condition `gauche < droite`.

L'invariant est ainsi conservé : si une solution non encore trouvée existe, elle
reste dans la zone courante. La largeur de cette zone diminue strictement à chaque
tour, ce qui prouve aussi la terminaison.

Dans une variante lecture/écriture, la preuve change. Après avoir lu les positions
`0..lecture`, le préfixe avant `ecriture` contient exactement les éléments gardés,
dans leur ordre original. Comme `ecriture` ne dépasse jamais `lecture`, une écriture
ne détruit pas une valeur qui reste à examiner.


## 4. Le prix du tri
Sur une liste déjà triée, O(n) temps et O(1) espace de travail. Si on doit la
trier, le total est O(n log n) dans le pire cas pour le tri Python. `sorted`
crée une copie. Les indices du tableau trié ne sont plus ceux de l'entrée.
Si le contrat exige les indices originaux, stocke les indices avec les valeurs,
ou utilise un dictionnaire : ce n'est pas le même contrat.

### Exemple illustré — compter aussi le tri

```text
liste non triée
      ↓ tri : O(n log n)
liste triée
      ↓ deux pointeurs : O(n)
réponse
```

Le coût total contient les deux étapes. Si la liste est déjà triée, la première
disparaît ; sinon elle ne doit pas être oubliée.




*Image — Ne pas oublier le travail effectué avant le parcours.*


## 5. Quand préférer cette technique ?
Les indices opposés conviennent à une propriété monotone liée à l'ordre,
ou à des comparaisons symétriques (palindrome). Lecture/écriture convient
à un filtrage sur place. Deux indices ne rendent pas automatiquement un code O(n) :
si l'on recommence le parcours entier à chaque tour, le coût peut être quadratique.

### Exemple illustré — vérifier les conditions avant de choisir

| Question | Oui | Non |
|---|---|---|
| les valeurs sont-elles triées ? | deux extrémités possibles | chercher une autre information |
| faut-il conserver les indices d’origine ? | attention au tri | le tri peut être acceptable |
| chaque mouvement élimine-t-il un cas ? | progression justifiable | technique mal adaptée |

Deux variables ne suffisent pas : il faut pouvoir expliquer chaque déplacement.


## Paire sur une liste triée

La fonction renvoie les indices dans la liste reçue, ou None. Elle suppose l’entrée triée ; la vérifier coûterait aussi O(n).


````python
def paire_triee(nums, cible, trace=False):
    gauche, droite = 0, len(nums)-1
    while gauche < droite:
        total = nums[gauche] + nums[droite]
        if trace:
            print(gauche, droite, total)
        if total == cible:
            return gauche, droite
        if total < cible:
            gauche += 1
        else:
            droite -= 1
    return None
assert paire_triee([1,2,4,7,11], 9, True) == (1,3)
assert paire_triee([3,3], 6) == (0,1)
assert paire_triee([3], 6) is None
````

### Trace visuelle : une comparaison, une élimination justifiée

Pour `[1, 2, 4, 7, 11]` et la cible `9` :

| Gauche | Droite | Valeurs | Somme | Conclusion sûre | Mouvement |
|---:|---:|---|---:|---|---|
| 0 | 4 | `1 + 11` | 12 | trop grand | droite recule |
| 0 | 3 | `1 + 7` | 8 | trop petit | gauche avance |
| 1 | 3 | `2 + 7` | 9 | cible trouvée | arrêt |

Le mouvement n’est pas un essai au hasard : grâce au tri, toute l’extrémité
abandonnée peut être éliminée sans perdre une solution possible.

Trace à expliquer : pourquoi déplacer droite sur 12 ? Pourquoi déplacer gauche sur 8 ? Sur une entrée non triée, ces éliminations ne sont plus justifiées.

### Palier A — Palindrome exact

Renvoie True si une chaîne se lit de la même manière dans les deux sens. Respecte espaces et casse. La chaîne vide est un palindrome.

<details><summary>Indice progressif</summary>

Compare les deux extrémités puis rapproche-les.

</details>


````python
def palindrome(texte):
    gauche, droite = 0, len(texte)-1
    while gauche < droite:
        if texte[gauche] != texte[droite]:
            return False
        gauche += 1
        droite -= 1
    return True
````

````python
assert palindrome("radar")
assert palindrome("")
assert not palindrome("Radar")
assert not palindrome("ab")
````

**Pourquoi cette solution fonctionne**

O(n) dans le pire cas et O(1) espace. Une discordance suffit à répondre False immédiatement.


### Palier B — Compacter les zéros

Déplace les zéros à la fin de la liste sur place, en gardant l’ordre des autres valeurs. Ne renvoie rien.

<details><summary>Indice progressif</summary>

Un indice lit tous les éléments, un autre désigne la prochaine position d’écriture.

</details>


````python
def deplacer_zeros(nums):
    ecriture = 0
    for lecture in range(len(nums)):
        if nums[lecture] != 0:
            nums[ecriture] = nums[lecture]
            ecriture += 1
    for i in range(ecriture, len(nums)):
        nums[i] = 0
````

````python
a = [0,1,0,3,12]
assert deplacer_zeros(a) is None
assert a == [1,3,12,0,0]
a = []
deplacer_zeros(a)
assert a == []
a = [0,0]
deplacer_zeros(a)
assert a == [0,0]
````

**Pourquoi cette solution fonctionne**

L’indice d’écriture ne dépasse jamais celui de lecture, donc on n’écrase pas une valeur encore nécessaire. Deux boucles successives : O(n), espace O(1).


### Palier C — Paires uniques par valeurs

Sur une liste triée, renvoie toutes les paires de valeurs distinctes au sens du résultat dont la somme vaut cible. [1,1,2,2,3,3], cible 4 donne [(1,3),(2,2)].

<details><summary>Indice progressif</summary>

Après avoir trouvé une paire, saute toutes les occurrences des deux valeurs utilisées.

</details>


````python
def paires_uniques(nums, cible):
    gauche, droite = 0, len(nums)-1
    resultat = []
    while gauche < droite:
        total = nums[gauche] + nums[droite]
        if total < cible:
            gauche += 1
        elif total > cible:
            droite -= 1
        else:
            a, b = nums[gauche], nums[droite]
            resultat.append((a,b))
            while gauche < droite and nums[gauche] == a:
                gauche += 1
            while gauche < droite and nums[droite] == b:
                droite -= 1
    return resultat
````

````python
assert paires_uniques([1,1,2,2,3,3],4) == [(1,3),(2,2)]
assert paires_uniques([], 3) == []
assert paires_uniques([2], 4) == []
````

**Pourquoi cette solution fonctionne**

Les indices avancent sans recul, y compris dans les boucles de saut : O(n). Espace O(p) pour p paires renvoyées, O(1) hors sortie.
