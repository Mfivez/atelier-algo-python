# 2 — Choisir une structure de données


## Fil conducteur

Le chapitre 1 a montré qu'une recherche répétée dans une liste peut coûter O(n)
à chaque fois. L'idée de ce chapitre est simple : conserver l'information dont
on aura besoin plus tard. Nous partirons de la liste, puis ajouterons une seule
capacité à la fois : présence, association, ordre inverse et ordre d'arrivée.


## 1. La liste : conserver un ordre

Une liste est une suite ordonnée. Chaque valeur possède une position appelée
**indice**. Lire une position connue, comme <code>valeurs[3]</code>, coûte O(1).
En revanche, chercher une valeur avec <code>x in valeurs</code> peut demander de
parcourir toute la liste : O(n).

Une liste convient lorsque l'ordre, les doublons ou l'accès par position sont
importants. Elle n'est pas toujours le meilleur choix pour répondre souvent à
la question « cette valeur a-t-elle déjà été vue ? ».

### Exemple illustré — une liste conserve positions et doublons

```text
indice :   0    1    2
valeur : [ A ][ B ][ A ]
```

Les deux `A` restent présents et leur position est connue. Lire l’indice 1 donne
directement `B`.


## 2. L'ensemble : retenir une présence

Un <code>set</code>, appelé **ensemble**, conserve des valeurs distinctes. Il ne
représente ni leur nombre d'occurrences ni une position précise. Son usage
typique est :

> Ai-je déjà rencontré cette valeur ?

Recherche et insertion coûtent O(1) en moyenne. « En moyenne » signifie que le
mécanisme de rangement répartit habituellement bien les valeurs, sans promettre
que chaque situation imaginable sera constante.

### Exemple illustré — un ensemble répond à « déjà vu ? »

```text
entrée :   A → B → A → C
ensemble : {A} → {A,B} → {A,B} → {A,B,C}
```

Le second `A` ne crée pas de nouvelle valeur. L’ensemble conserve la présence,
mais pas les doublons de l’entrée.


### Le hachage, sans boîte noire

Le **hachage** transforme une clé en un nombre servant à choisir une zone de
stockage. Deux clés peuvent choisir la même zone : c'est une **collision**.
Python les compare alors pour les distinguer. Lorsque les collisions restent
peu nombreuses, la recherche est O(1) en moyenne.

Une clé doit garder le même hachage pendant son stockage. Les nombres, chaînes
et tuples de valeurs hachables conviennent généralement. Une liste est mutable :
son contenu peut changer, elle ne peut donc pas servir directement de clé.




*Le hachage choisit une case ; la comparaison des clés résout les collisions éventuelles.*

## 3. Le dictionnaire : associer une clé à une information

Un <code>dict</code>, appelé **dictionnaire**, associe une clé à une valeur.
La valeur peut représenter un compteur, un indice, un parent ou toute autre
information utile.

Pour compter « a, b, a », on construit progressivement :

| valeur lue | dictionnaire après la lecture |
|---|---|
| a | {a: 1} |
| b | {a: 1, b: 1} |
| a | {a: 2, b: 1} |

Le dictionnaire répond donc à une question plus riche que l'ensemble :
non seulement « a-t-on vu a ? », mais aussi « combien de fois ? ».

### Exemple illustré — une clé mène à une information

```text
"Ada"  ──→  3 exercices
"Lina" ──→  5 exercices
"Noé"  ──→  2 exercices
```

Le nom est la **clé** utilisée pour retrouver la **valeur** associée. Une clé
n’apparaît qu’une fois dans le dictionnaire.


## 4. Exemple : mémoriser les valeurs précédentes

Pour chercher deux nombres dont la somme vaut une cible, on pourrait essayer
toutes les paires en O(n²). À la place, lors de la lecture de x, on calcule le
complément <code>cible - x</code> et on demande s'il a déjà été vu.

Le dictionnaire contient uniquement des indices antérieurs. Vérifier avant
d'insérer x empêche de réutiliser deux fois la même position.

### Exemple illustré — chercher le complément déjà vu

Pour la cible 9 :

| Valeur actuelle | Complément | Valeurs déjà vues | Résultat |
|---:|---:|---|---|
| 2 | 7 | `{}` | mémoriser 2 |
| 7 | 2 | `{2}` | 2 trouvé, donc `2 + 7 = 9` |

On ne cherche que dans le passé du parcours, ce qui évite de tester toutes les
paires.


````python
def deux_sommes(nums, cible):
    vus = {}
    for i, x in enumerate(nums):
        complement = cible - x
        if complement in vus:
            return vus[complement], i
        vus[x] = i
    return None

assert deux_sommes([2, 7, 11], 9) == (0, 1)
assert deux_sommes([3, 3], 6) == (0, 1)
assert deux_sommes([3], 6) is None
````

### Exemple visuel : le dictionnaire se construit au fil du parcours

Avec `nums = [2, 7, 11]` et `cible = 9` :

| Étape | `x` | Complément recherché `9 - x` | `vus` avant le test | Décision |
|---:|---:|---:|---|---|
| 0 | 2 | 7 | `{}` | 7 absent, mémoriser `2: 0` |
| 1 | 7 | 2 | `{2: 0}` | 2 présent, rendre `(0, 1)` |

Le dictionnaire ne stocke pas « la réponse ». Il stocke l’information qui permettra
de reconnaître rapidement une réponse plus tard : valeur déjà vue → position.

Chaque position est lue une fois. Les recherches dans le dictionnaire coûtent
O(1) en moyenne : le temps total est O(n) en moyenne. Le dictionnaire peut
contenir jusqu'à n clés : O(n) mémoire de travail.

Nous avons échangé de la mémoire contre du temps. Ce compromis apparaît souvent
en algorithmique.


## 5. La pile : dernier entré, premier sorti

Imagine une pile d'assiettes : la dernière posée est la première retirée.
Cet ordre est nommé **LIFO**, pour « Last In, First Out », c'est-à-dire
« dernier entré, premier sorti ».

Une liste Python convient avec <code>append</code> pour empiler et
<code>pop()</code> pour dépiler. Une pile est utile pour revenir en arrière,
gérer des appels ou fermer des parenthèses dans l'ordre inverse de leur ouverture.

### Exemple illustré — la pile retire le dernier élément ajouté

```text
push A       push B       pop
┌───┐        ┌───┐        ┌───┐
│ A │        │ B │  ←     │ A │
└───┘        ├───┤        └───┘
             │ A │
             └───┘
```

`B` est entré en dernier : il sort en premier.


## 6. La file : premier entré, premier sorti

Imagine une file d'attente : la première personne arrivée est la première servie.
Cet ordre est nommé **FIFO**, pour « First In, First Out ».

<code>collections.deque</code> fournit <code>append</code> pour ajouter à droite
et <code>popleft</code> pour retirer à gauche, en O(1). Une liste avec
<code>pop(0)</code> doit décaler les éléments restants et coûte O(n).

La file sera utilisée au chapitre 7 pour explorer un graphe couche après couche.

### Exemple illustré — la file respecte l’ordre d’arrivée

```text
sortie ← [ A ][ B ][ C ] ← entrée
           ↑
      prochain retiré
```

`A` attend depuis le plus longtemps : il sort avant `B` et `C`.


## 7. Choisir en partant de la question

| Question | Structure candidate |
|---|---|
| Ai-je besoin de l'ordre ou des positions ? | liste |
| Ai-je déjà vu cette valeur ? | ensemble |
| Quelle information correspond à cette clé ? | dictionnaire |
| Dois-je traiter le dernier élément ajouté ? | pile |
| Dois-je traiter le premier élément arrivé ? | file |

Ce tableau propose un point de départ, pas une réponse automatique. Le contrat,
les opérations nécessaires et leur coût doivent confirmer le choix.

### Exemple illustré — même donnée, questions différentes

| Question | Information importante | Structure candidate |
|---|---|---|
| quel est le 3e élément ? | position | liste |
| cette valeur existe-t-elle ? | présence | ensemble |
| combien de fois apparaît-elle ? | fréquence par valeur | dictionnaire |
| quel élément fermer en premier ? | dernier ouvert | pile |

La structure est choisie à partir de la question, pas du nom de l’exercice.




*Chaque structure est reliée à l’information ou à l’ordre qu’elle doit conserver.*

### Palier A — Valeurs uniques dans l'ordre

Renvoie une nouvelle liste contenant la première occurrence de chaque valeur
hachable. Ne modifie pas l'entrée.


````python
def uniques(valeurs):
    vus = set()
    resultat = []
    for valeur in valeurs:
        if valeur not in vus:
            vus.add(valeur)
            resultat.append(valeur)
    return resultat
````

````python
assert uniques([3, 1, 3, 2, 1]) == [3, 1, 2]
assert uniques([]) == []
````

**Pourquoi cela fonctionne.** L'ensemble mémorise exactement les valeurs déjà
ajoutées. La liste résultat conserve leur ordre de première apparition.
Temps O(n) en moyenne, mémoire O(u) pour u valeurs distinctes.


### Palier B — Délimiteurs correctement imbriqués

Vérifie les couples (), [] et {}. Ignore les autres caractères. Chaque fermant
doit correspondre au dernier ouvrant non encore fermé.


````python
def delimiteurs_valides(texte):
    pile = []
    correspondance = {')': '(', ']': '[', '}': '{'}
    for caractere in texte:
        if caractere in "([{":
            pile.append(caractere)
        elif caractere in correspondance:
            if not pile or pile.pop() != correspondance[caractere]:
                return False
    return not pile
````

````python
assert delimiteurs_valides("a{b[c(d)]}")
assert not delimiteurs_valides("([)]")
assert not delimiteurs_valides("(")
````

**Pourquoi cela fonctionne.** La pile contient les ouvrants encore en attente.
Son sommet est le seul ouvrant qui puisse être fermé maintenant. Temps O(n),
mémoire O(n) dans le pire cas.


### Palier C — Regrouper les positions

Renvoie un dictionnaire qui associe chaque valeur hachable à la liste de ses
indices. Exemple : ["a", "b", "a"] donne {"a": [0, 2], "b": [1]}.


````python
def positions_par_valeur(valeurs):
    resultat = {}
    for indice, valeur in enumerate(valeurs):
        if valeur not in resultat:
            resultat[valeur] = []
        resultat[valeur].append(indice)
    return resultat
````

````python
assert positions_par_valeur(["a", "b", "a"]) == {"a": [0, 2], "b": [1]}
assert positions_par_valeur([]) == {}
````

**Pourquoi cela fonctionne.** Chaque clé possède sa propre liste créée lors de
sa première apparition. Tous les indices sont ajoutés dans l'ordre du parcours.
Temps O(n) en moyenne et O(n) pour la sortie.


## Questions de compréhension

1. Quelle information un ensemble conserve-t-il, et laquelle perd-il ?
2. Pourquoi une liste ne peut-elle pas servir de clé de dictionnaire ?
3. Dans <code>deux_sommes</code>, pourquoi cherche-t-on avant d'insérer ?
4. Quelle différence d'ordre sépare une pile d'une file ?


**Éléments de réponse.** Un ensemble conserve la présence, pas l'ordre ni le
nombre d'occurrences. Une liste est mutable et son hachage ne serait pas stable.
Chercher avant l'insertion évite de réutiliser le même indice. La pile traite le
dernier arrivé ; la file traite le premier arrivé.
