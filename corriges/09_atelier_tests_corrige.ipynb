## 1. Une méthode reproductible
1. Reformuler entrée et sortie. Demander si une liste est triée, si les valeurs peuvent
   être négatives, si les doublons comptent et si la modification sur place est autorisée.
2. Écrire un petit exemple et sa réponse à la main.
3. Proposer une solution correcte même simple. Ne pas choisir une technique au seul
   mot-clé présent dans l'énoncé.
4. Identifier le travail répété et les informations réutilisables.
5. Énoncer l'invariant ou la règle qui justifie l'optimisation.
6. Tester vide, un élément, doublons, absence de réponse, signes et égalités aux bornes.
7. Analyser le coût en distinguant entrée, travail et sortie.

### Exemple illustré — une résolution suit un pipeline stable

```text
contrat → petit exemple → solution simple → tests
        → coût → optimisation justifiée → nouveaux tests
```

Chaque flèche dépend de l’étape précédente. Optimiser avant d’avoir un contrat et
une solution correcte retire les points de comparaison utiles.




*La solution simple et sa justification précèdent les tests puis l’optimisation.*


## 2. Une grille de choix, pas une recette automatique
| Besoin | Technique candidate | Question à vérifier |
|---|---|---|
| Portion contiguë de taille k | Fenêtre fixe | L’état se met-il à jour en O(1) ? |
| Portion contiguë sous contrainte | Fenêtre variable | La réduction préserve-t-elle les possibilités ? |
| Paire dans une liste triée | Deux indices opposés | Le tri est-il déjà acquis ? |
| Présence ou fréquence | Set / dictionnaire | Les valeurs sont-elles hachables ? |
| Toutes les combinaisons | Backtracking | Peut-on élaguer ? Quelle taille de sortie ? |
| Format textuel | Regex | Format et sens sont-ils distincts ? |
| Imbrication | Pile | Faut-il un vrai parseur ? |
| Chemin minimal sans poids | BFS | Les arêtes ont-elles toutes le même coût ? |

### Exemple illustré — les propriétés conduisent au candidat

Énoncé : « plus longue sous-chaîne sans doublon ».

| Propriété observée | Conséquence |
|---|---|
| sous-chaîne contiguë | penser à une fenêtre |
| taille non fixée | fenêtre variable |
| doublon rend la zone invalide | ensemble ou fréquences |

La technique résulte de plusieurs propriétés concordantes, pas d’un mot-clé isolé.


### Exemple visuel : transformer l’énoncé en décision

Énoncé : « trouver la plus longue sous-chaîne sans caractère répété ».

| Question posée dans l’ordre | Réponse observée | Conséquence |
|---|---|---|
| La réponse est-elle contiguë ? | oui, c’est une sous-chaîne | considérer une fenêtre |
| La taille est-elle fixée ? | non | fenêtre variable |
| Quand la fenêtre devient-elle invalide ? | à l’arrivée d’un doublon | réduire depuis la gauche |
| Quelle information faut-il maintenir ? | caractères présents | ensemble ou fréquences |
| Pourquoi le coût peut-il être O(n) ? | chaque borne avance au plus n fois | justifier avant de coder |

Le mot « sous-chaîne » ne suffit pas à choisir une technique. La décision vient de
la suite de propriétés vérifiées dans la colonne centrale.

## 4. Concevoir des tests, pas seulement des exemples

Un test possède une entrée, un résultat attendu — l'**oracle** — et une comparaison.
Une bonne petite suite combine plusieurs sources de cas :

- **partitions** : une entrée représentative de chaque comportement, par exemple
  réponse présente ou absente ;
- **bornes** : vide, un élément, juste avant, sur et juste après une limite ;
- **cas structurels** : doublons, ordre inversé, cycle, caractères particuliers ;
- **propriétés** : le résultat est trié, conserve la somme, contient exactement les
  éléments attendus ou respecte un invariant ;
- **comparaison différentielle** : sur de petites entrées, comparer l'optimisation
  à une référence simple ;
- **aléatoire reproductible** : fixer une graine et conserver tout cas qui révèle
  un défaut comme test de régression.

Un exemple heureux montre seulement qu'un chemin fonctionne. Un test négatif
vérifie qu'une entrée interdite est refusée. Les tests peuvent révéler une erreur,
mais un nombre fini de tests ne prouve généralement pas l'absence d'erreur : la
preuve de correction et les tests jouent des rôles complémentaires.

Dans une cellule de test, sépare mentalement trois moments : préparer les données,
appeler la fonction, vérifier le résultat et les éventuels effets de bord. Si une
fonction modifie une liste, contrôle à la fois sa valeur de retour et l'état final
de cette liste.

### Exemple illustré — une petite matrice de tests

Pour une recherche de paire :

| Famille | Entrée | Attendu |
|---|---|---|
| réponse présente | `[2, 7]`, cible 9 | une paire |
| réponse absente | `[2, 4]`, cible 9 | aucune paire |
| doublons | `[3, 3]`, cible 6 | paire valide |
| trop court | `[5]`, cible 10 | aucune paire |

Chaque ligne cible un risque différent ; quatre variantes du cas heureux seraient
moins utiles.




*Un jeu de tests utile couvre plusieurs familles de risques au lieu d’accumuler des exemples similaires.*

## 5. Mesurer sans confondre avec démontrer
`timeit` mesure une exécution dans un environnement donné ; Big O décrit une
croissance asymptotique. Une mesure isolée ne prouve pas O(n). Compare les mêmes
entrées et évite de mesurer un affichage ou la génération des données. Une solution
qui paraît rapide sur n=10 peut devenir mauvaise sur n=100000.

### Exemple illustré — mesurer n’est pas démontrer

| Observation locale | Conclusion permise | Conclusion interdite |
|---|---|---|
| 1 ms pour `n=1 000` | cette exécution a pris 1 ms | toujours O(1) |
| temps environ doublé quand `n` double | indice d’une croissance linéaire | preuve pour toute entrée |

La mesure compare des exécutions réelles. L’analyse explique la croissance à partir
des opérations du programme.


### Tester la correction, observer la performance

Un test de correction doit produire la même décision sur une machine lente ou
rapide. Un benchmark mesure au contraire une durée dépendante de l'environnement.
Pour comparer deux implémentations, utilise les mêmes données, plusieurs répétitions
et évite d'inclure la génération de l'entrée ou les affichages si ce n'est pas ce
que tu veux mesurer.

Une courbe empirique peut suggérer une croissance, jamais remplacer le comptage des
opérations. Inversement, deux fonctions de même classe asymptotique peuvent avoir
des constantes très différentes. On choisit une solution avec les deux informations :
le modèle théorique pour prévoir le changement d'échelle, la mesure pour comprendre
le comportement concret dans un contexte donné.


## Vérifier une optimisation avec une référence

Pour de petites entrées, une version naïve est une bonne référence de correction.


````python
import random

def reference_somme_max(nums, k):
    return max(sum(nums[i:i+k]) for i in range(len(nums)-k+1))

def optimisee_somme_max(nums, k):
    courant = sum(nums[i] for i in range(k))
    meilleur = courant
    for i in range(k,len(nums)):
        courant += nums[i]-nums[i-k]
        meilleur = max(meilleur,courant)
    return meilleur

rng = random.Random(2026)
for n in range(1,20):
    for _ in range(10):
        nums = [rng.randint(-10,10) for _ in range(n)]
        for k in range(1,n+1):
            assert reference_somme_max(nums,k) == optimisee_somme_max(nums,k)
print("1900 comparaisons de fenêtres validées")
````

Les tests aléatoires à graine fixe sont reproductibles et couvrent les signes et différentes tailles de fenêtre. Ils complètent la justification, sans la remplacer.


## Une mesure locale et modeste

Exécute plusieurs fois et observe le bruit. Aucun seuil de vitesse ne détermine la correction.


````python
from timeit import repeat
nums = list(range(800))
k = 160
naif = min(repeat(lambda: reference_somme_max(nums,k), number=3, repeat=3))/3
rapide = min(repeat(lambda: optimisee_somme_max(nums,k), number=3, repeat=3))/3
print(f"naïf : {naif:.6f} s ; fenêtre : {rapide:.6f} s")
````

Pour k fixé, les deux méthodes ont une croissance linéaire en n, mais des constantes différentes. Ici k est assez grand pour montrer le travail évité ; cela ne suffit pas à établir une loi asymptotique.

### Palier A — Premier caractère unique

Renvoie l’indice du premier caractère apparaissant exactement une fois, ou −1. Respecte la casse et les espaces.

<details><summary>Indice progressif</summary>

Compte les fréquences, puis parcours dans l’ordre original.

</details>


````python
def premier_unique(texte):
    comptes = {}
    for c in texte:
        comptes[c] = comptes.get(c,0)+1
    for i,c in enumerate(texte):
        if comptes[c] == 1:
            return i
    return -1
````

````python
assert premier_unique("swiss") == 1
assert premier_unique("aabb") == -1
assert premier_unique("") == -1
assert premier_unique("Aa") == 0
````

**Pourquoi cette solution fonctionne**

Deux traversées successives : O(n) en moyenne, pas O(n²). Mémoire O(u) pour u caractères distincts.


### Palier B — Meilleure moyenne de k mesures

Renvoie (indice_de_depart, moyenne) de la meilleure fenêtre de k valeurs. En cas d’égalité, garde la première. Refuse k hors 1..n.

<details><summary>Indice progressif</summary>

Maximiser la moyenne revient à maximiser la somme car k est le même pour toutes les fenêtres. Utilise >, pas >=, pour préserver le premier ex æquo.

</details>


````python
def meilleure_moyenne(nums, k):
    if not 1 <= k <= len(nums):
        raise ValueError("k invalide")
    courant = sum(nums[i] for i in range(k))
    meilleur, debut = courant, 0
    for droite in range(k,len(nums)):
        courant += nums[droite]-nums[droite-k]
        if courant > meilleur:
            meilleur, debut = courant, droite-k+1
    return debut, meilleur/k
````

````python
assert meilleure_moyenne([2,1,5,1,3,2],3) == (2,3.0)
assert meilleure_moyenne([1,1,1],2) == (0,1.0)
assert meilleure_moyenne([-4,-2,-3],2) == (1,-2.5)
````

**Pourquoi cette solution fonctionne**

O(n) temps, O(1) travail et sortie. Ces tests emploient des entiers ; avec des flottants, l’addition glissante peut accumuler des erreurs d’arrondi et les égalités nécessitent une politique explicite.


### Palier C — Permutations distinctes

Pour une liste de valeurs distinctes et hachables, renvoie toutes les permutations sous forme de listes. Refuse les doublons avec ValueError. Liste vide → [[]].

<details><summary>Indice progressif</summary>

Un tableau utilisé[] empêche de prendre deux fois la même position. Annule à la fois le chemin et le marquage.

</details>


````python
def permutations(valeurs):
    if len(set(valeurs)) != len(valeurs):
        raise ValueError("valeurs distinctes requises")
    utilises = [False]*len(valeurs)
    chemin, resultat = [], []
    def visiter():
        if len(chemin) == len(valeurs):
            resultat.append(chemin.copy())
            return
        for i, valeur in enumerate(valeurs):
            if not utilises[i]:
                utilises[i] = True
                chemin.append(valeur)
                visiter()
                chemin.pop()
                utilises[i] = False
    visiter()
    return resultat
````

````python
assert permutations([]) == [[]]
assert permutations([1,2]) == [[1,2],[2,1]]
p = permutations([1,2,3])
assert len(p) == 6
assert len({tuple(x) for x in p}) == 6
assert all(sorted(x) == [1,2,3] for x in p)
````

**Pourquoi cette solution fonctionne**

La fonction restaure deux états partagés. O(n·n!) temps et sortie, O(n) travail : chaque solution coûte n à copier, et les boucles sur n candidats à chaque nœud restent dans cette borne. N’essaie pas des tailles élevées pour une simple démonstration.
